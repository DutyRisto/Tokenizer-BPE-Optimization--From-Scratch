```python
#step  2

import regex as re 
from collections import Counter
from pathlib import Path

def pretokenizer(input_path:str,
                 special_tokens:str|None = None) :
    
    Pat = r"""'(?:[sdmt]|ll|ve|re)| ?\p{L}+| ?\p{N}+| ?[^\s\p{L}\p{N}]+|\s+(?!\S)|\s+"""
    text=Path(input_path).read_text(encoding='utf-8')

    if special_tokens:
        motif ='|'.join(
            re.escape(token)
            for token in sorted(special_tokens,key=len,reverse=True)
        )
        chunks=re.split(motif,text)
    else:
        chunks=[text]

    patern=re.compile(Pat)
    counts=Counter()
    for chunk in chunks:
        for m in patern.finditer(chunk):
            token=m.group()
            counts[token]+=1
    return counts
pretokenizer(input_path=r"C:\Users\lenovo\Documents\Cours LLM Stanford\code_apprentissage\sample.txt")

```

```python
# Tant que le vocabulaire est trop petit :

#     1. Compter toutes les paires.
#     2. Trouver la meilleure paire.
#     3. Ajouter la fusion à merges.
#     4. Ajouter le nouveau token à vocab.
#     5. Mettre à jour les séquences.
```

```python

def train_bpe(
    input_path: str,
    vocab_size: int,
    special_tokens: list[str],
):
    # 1. Initialiser le vocabulaire
    vocab = {
        i: bytes([i])
        for i in range(256)
    }

    # Ajouter les tokens spéciaux
    special_tokens = list(dict.fromkeys(special_tokens))

    for token in special_tokens:
        vocab[len(vocab)] = token.encode("utf-8")

    if vocab_size < len(vocab):
        raise ValueError("vocab_size est trop petit")

    # 2. Pré-tokeniser
    counts = pretokenizer(
        input_path,
        special_tokens
    )

    # 3. Convertir en octets
    byte_counts = convert_to_bytes(counts)

    # Liste ordonnée des fusions
    merges = []

    # 4. Boucle d'apprentissage
    while len(vocab) < vocab_size:

        # Compter les paires
        pair_counts = count_pairs(byte_counts)

        if not pair_counts:
            break

        # Trouver la paire la plus fréquente
        best_pair = max(
            pair_counts,
            key=lambda p: (pair_counts[p], p)
        )

        # Créer le nouveau token
        new_token = best_pair[0] + best_pair[1]

        # Ajouter la fusion
        merges.append(best_pair)

        # Ajouter le nouveau token au vocabulaire
        vocab[len(vocab)] = new_token

        # Actualiser le corpus
        byte_counts = update_sequences(
            byte_counts,
            best_pair
        )

    return vocab, merges
```

```python
import os 
from typing  import BinaryIO


def find_chunk_boundaries(
        file :BinaryIO,
        desired_num_chunks : int ,
        split_special_token : bytes | None=None,
        *args , **kwargs
    )->list[int]: 


    assert split_special_token is None or isinstance(split_special_token,bytes)# Must represent special token as a bytes

    # Cursor position  
    # total size  like len but len not a function of BinaryIO format or ByteIO
    file.seek(0,os.SEEK_END)#end  [move cursor at the end ]
    file_size= file.tell()# position (where the cursor) # give the lenght of octets in the file 
    print(file_size)
    file.seek(0)

    # initialisation
    chunk_size= file_size // desired_num_chunks # octet by chunk
    print(chunk_size)
    chunk_boundaries=[i * chunk_size for i in range(desired_num_chunks+1)]
    chunk_boundaries[-1]=file_size
    # print(chunk_boundaries)
    # return chunk_boundaries

    mini_chunk_size = 4096 # read ahead by 4k bytes at a time 

    for bi in range(1,len(chunk_boundaries) -1):
        start=chunk_boundaries[bi] # initial position 
        file.seek(start)
        print(file.seek(start)) 

        while True : 
            # read
        
            mini_chunk= file.read(mini_chunk_size)
            print("mini",mini_chunk)
            
            if mini_chunk == b'': # end of the file

                chunk_boundaries[bi]=file_size
                break
            # found a special token 

            found_at=mini_chunk.find(split_special_token) # return a position of the match
            #  find -> for bytes  return int >=0 or -1 if not found 
        

            if  found_at !=-1: 
                chunk_boundaries[bi] = start + found_at
                break
            start +=mini_chunk_size

        return sorted(set(chunk_boundaries))

## usage




        # end =chunk_boundaries[bi+1]


with open("sample.txt", "rb") as f:
    result = find_chunk_boundaries(
        file=f,
        desired_num_chunks=4,
        split_special_token=b"<|endoftext|>"
    )

# # print(result)
# a=find_chunk_boundaries(fake_file,4)



```
